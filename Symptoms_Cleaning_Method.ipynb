```python
import pandas as pd

# Load your CSV file
df = pd.read_csv("C:\\Users\\Bhavya Saini\\OneDrive\\Documents\\Desktop\\Research\\HealthData-WestDelhi (1).csv")

# Function to process each cell and create a structured DataFrame
def process_symptoms(symptom_str):
    if pd.isna(symptom_str):
        return []
    symptoms = symptom_str.split(",")
    processed_symptoms = []
    for symptom in symptoms:
        if '**' in symptom:
            name, duration = symptom.split('**')
            processed_symptoms.append((name.strip(), duration.strip()))
        else:
            processed_symptoms.append((symptom.strip(), ''))
    return processed_symptoms

# Process each row and collect results
processed_data = [process_symptoms(symptom_str) for symptom_str in df['symptoms']]

# Create a new DataFrame from the processed data
max_columns = max(len(row) for row in processed_data)
columns = sum([[f'Symptom{i+1}', f'Duration{i+1}'] for i in range(max_columns)], [])

# Flatten the processed data into rows
flattened_data = []
for row in processed_data:
    flattened_row = sum(row, ())
    flattened_data.append(flattened_row + ('',) * (2 * max_columns - len(flattened_row)))

# Create the final DataFrame
result_df = pd.DataFrame(flattened_data, columns=columns)

# Replace the symptom column in df with the new structured columns in result_df
df = df.drop(columns=['symptoms'])
df = pd.concat([df, result_df], axis=1)

df['Symptom1'] = df['Symptom1'].str.replace('Other :: ', '').str.strip()
df['Symptom2'] = df['Symptom2'].str.replace('Other :: ', '').str.strip()
df['Symptom3'] = df['Symptom3'].str.replace('Other :: ', '').str.strip()
df['Symptom4'] = df['Symptom4'].str.replace('Other :: ', '').str.strip()
df['Symptom5'] = df['Symptom5'].str.replace('Other :: ', '').str.strip()
df['Symptom6'] = df['Symptom6'].str.replace('Other :: ', '').str.strip()
df['Symptom7'] = df['Symptom7'].str.replace('Other :: ', '').str.strip()
df['Symptom8'] = df['Symptom8'].str.replace('Other :: ', '').str.strip()
df['Symptom9'] = df['Symptom9'].str.replace('Other :: ', '').str.strip()
df['Symptom10'] = df['Symptom10'].str.replace('Other :: ', '').str.strip()

# Extract unique values from each Symptom column
unique1 = df['Symptom1'].unique()
unique2 = df['Symptom2'].unique()
unique3 = df['Symptom3'].unique()
unique4 = df['Symptom4'].unique()
unique5 = df['Symptom5'].unique()
unique6 = df['Symptom6'].unique()
unique7 = df['Symptom7'].unique()
unique8 = df['Symptom8'].unique()
unique9 = df['Symptom9'].unique()
unique10 = df['Symptom10'].unique()

# Create individual DataFrames from the unique arrays
unique_df1 = pd.DataFrame(unique1, columns=['Unique Symptoms'])
unique_df2 = pd.DataFrame(unique2, columns=['Unique Symptoms'])
unique_df3 = pd.DataFrame(unique3, columns=['Unique Symptoms'])
unique_df4 = pd.DataFrame(unique4, columns=['Unique Symptoms'])
unique_df5 = pd.DataFrame(unique5, columns=['Unique Symptoms'])
unique_df6 = pd.DataFrame(unique6, columns=['Unique Symptoms'])
unique_df7 = pd.DataFrame(unique7, columns=['Unique Symptoms'])
unique_df8 = pd.DataFrame(unique8, columns=['Unique Symptoms'])
unique_df9 = pd.DataFrame(unique9, columns=['Unique Symptoms'])
unique_df10 = pd.DataFrame(unique10, columns=['Unique Symptoms'])

# Combine all the DataFrames into a single DataFrame
combined_df = pd.concat([unique_df1, unique_df2, unique_df3, unique_df4, unique_df5, unique_df6, unique_df7, unique_df8, unique_df9, unique_df10])

# Remove duplicate values to get the final unique values
final_unique_df = combined_df.drop_duplicates().reset_index(drop=True)

# Define the path to the output Excel file
output_file_path = r"C:\Users\Bhavya Saini\OneDrive\Documents\Desktop\Research\Unique_Symptoms.xlsx"

# Save the final DataFrame to an Excel file
final_unique_df.to_excel(output_file_path, index=False)

print(f"Final unique values have been saved to {output_file_path}")

```
